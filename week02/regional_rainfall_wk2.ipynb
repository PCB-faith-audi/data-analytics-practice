```python
import numpy as np

regions = ["Central", "Coast", "Western"]
rain = np.array([[12, 0, 5, 30],
                 [8, 2, 0, 15],
                 [20, 10, 12, 40]])

totals = rain.sum(axis=1)
share = (totals / rain.sum() * 100).round(1)
driest = regions[np.argmin(totals)]
wettest = regions[np.argmax(totals)]
dry_weeks = (rain == 0).sum(axis=1)

for name, t, s, d in zip(regions, totals, share, dry_weeks):
    print(f"{name:<8} total={t:>3} mm share={s:>5}% dry weeks={d}")

print(f"Wettest: {wettest} | Driest: {driest}")
#
```
