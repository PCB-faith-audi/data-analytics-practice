```python
ages = ["23", "31", "unknown", "45", "29", "-5", "", "38", "200"]
print(ages)

def to_int_or_none(value):
  """Strip spaces and convert to int; return None if conversion fails."""
  try:
    return int(value.strip())
  except (ValueError, AttributeError):
    return None

print(to_int_or_none("23"))
print(to_int_or_none("unknown"))
print(to_int_or_none(""))

converted_ages = [to_int_or_none(age) for age in ages]
print(converted_ages)

filtered_ages = [age for age in converted_ages if age is not None and 15 <= age <= 100]
print(filtered_ages)

unparseable = sum(1 for age in converted_ages if age is None)
parsed = [age for age in converted_ages if age is not None]
out_of_range = len(parsed) - len(filtered_ages)

print(f"filtered ages: {len(filtered_ages)}")
print(f"unparseable: {unparseable}")
print(f"Out of range: {out_of_range}")
print(f"Total checked: {len(ages)}")
```

### Why is an age of 200 a validity problem even though it converts?

Successfully converting `"200"` to an integer only proves it is **well-formed**
(a valid number). It does not prove it is **valid** (a plausible real-world value).
Python's `int()` checks *format*, not *meaning*. No human is 200 years old, so the
value is almost certainly a typo (e.g., 20 or 2.00) or a data-entry error.

Data cleaning therefore needs two layers:
1. **Parsing check:** can the text become a number? (handled by `try/except`)
2. **Validity (range) check:** is the number realistic for the field? (handled by the 15–100 filter)

If we skipped the second check, bad values such as 200 and -5 would distort
statistics like the mean age and lead to wrong conclusions.
