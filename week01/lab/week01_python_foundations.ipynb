```python
# Sales calculator practice
trips = 18
avg_fare = 220.0
fuel_cost = 1100.0
rider_name = "Kevin"

gross_income = trips * avg_fare
net_income = gross_income - fuel_cost
margin_pct = net_income / gross_income * 100

print(f"{rider_name}: gross KES {gross_income:.2f},"
    f"net KES {net_income:,.2f}, marin {margin_pct:.1f}%")
```

```python
# Inventory alerts with dictionaries and loops
inventory = [
    {"name": "cakes", "stock": 30, "reorder_level": 37},
    {"name": "wheat flour", "stock": 20, "reorder_level": 50},
    {"name": "milk", "stock": 50, "reorder_level": 10},
    {"name": "oil", "stock": 30, "reorder_level": 200},
]

def restock_report(items):
  """Print stock status and return a list of products to reorder."""
  to_reorder = []
  for item in items:
    if item["stock"] < item["reorder_level"]:
      status = "REORDER"
      to_reorder.append(item["name"])
    else:
      status = "OK"
    print(f"{item['name']:<6} stock={item['stock']:>3} {status}")
  return to_reorder
 
print(restock_report(inventory))
```
