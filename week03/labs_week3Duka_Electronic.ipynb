```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(2026)
n = 1000
products = {"Phone": 15000, "Laptop": 65000, "Headphones": 2500,
            "Tablet": 28000, "Charger": 1200}

sales = pd.DataFrame({
    "order_id": range(1001, 1001 + n),
    "order_date": pd.to_datetime("2026-01-01")
              + pd.to_timedelta(rng.integers(0, 180, n), unit="D"),
    "branch": rng.choice(["Nairobi", "Mombasa", "Kisumu", "Nakuru", "Eldoret"], n,
                         p=[0.4, 0.2, 0.15, 0.15, 0.1]),
    "product":rng.choice(list(products), n),
    "quantity": rng.integers(1, 6, n),
    "channel": rng.choice(["In-store", "Online", "Phone order"], n,
                          p=[0.55, 0.35, 0.10]),

})

sales["unit_price"] = sales["product"].map(products)
sales.to_csv("duka_sales.csv", index=False)
```

```python
# Lab 3.2: Answer Ten Manager Questions
import pandas as pd

# Load the dataset to ensure 'df' is defined in this session
df = pd.read_csv("duka_sales.csv", parse_dates=["order_date"])

# Lab 3.2 Answer Ten Manager Questions

# Task 1: add a revenue column, then count Nairobi orders
df["revenue"] = df["quantity"] * df["unit_price"]

nairobi_orders = (df["branch"] == "Nairobi").sum()
print("Nairobi orders:", nairobi_orders)

# Task 2: online laptop orders
online_laptops = ((df["channel"] == "Online") & (df["product"] == "Laptop")).sum()
print("Online laptop orders:", online_laptops)

# Task 3: total revenue from Mombasa and Kisumu combined
coastal_lake = df[df["branch"].isin(["Mombasa", "Kisumu"])]
total_revenue = coastal_lake["revenue"].sum()

print(f"Mombasa + Kisumu revenue: KES {total_revenue:,}")

# Task 4: ten highest-revenue orders
top10 = df.nlargest(10, "revenue")[["order_id", "branch", "product", "revenue"]]
print(top10)

# Task 5: orders with quantity of four or more
big_orders = (df["quantity"] >= 4).sum()
print("Orders with quantity >= 4:", big_orders)

# Task 6: share of revenue from phone orders
phone_revenue = df.loc[df["channel"] == "Phone order", "revenue"].sum()
total_revenue = df["revenue"].sum()
phone_share = phone_revenue / total_revenue

print(f"Phone order revenue: KES {phone_revenue:,}")
print(f"Total revenue: KES {total_revenue:,}")
print(f"Phone order share: {phone_share:.1%}")

# Task 7: orders placed in March 2026
march_orders = df[(df["order_date"].dt.month == 3) & (df["order_date"].dt.year == 2026)]

print("March 2026 orders:", len(march_orders))
print(march_orders[["order_id", "order_date", "branch", "product", "revenue"]].head(10))

# Task 8: question 2 using .query()
online_laptops_query = len(df.query("channel == 'Online' and product == 'Laptop'"))
print("Online laptop orders (query):", online_laptops_query)

# Confirm it matches the Task 2 answer
print("Same answer as Task 2:", online_laptops_query == online_laptops)

# Task 9: average order revenue, in-store vs online
avg_revenue = df[df["channel"].isin(["In-store", "Online"])].groupby("channel")["revenue"].mean()

print(avg_revenue.round(2))

# Task 10: save March orders to CSV without the index
march_orders.to_csv("march_orders.csv", index=False)

print("march_orders.csv saved with", len(march_orders), "rows")

check = pd.read_csv("march_orders.csv", parse_dates=["order_date"])

print(check.shape)
print(check.columns.tolist())
print(check.head())
```

```python
# Lab 3.3: Feature Engineering Basics

# Task 1: month name and weekday name columns
df["month"] = df["order_date"].dt.month_name()
df["weekday"] = df["order_date"].dt.day_name()

print(df[["order_date", "month", "weekday"]].head())

# Task 2: is_weekend as a boolean column
df["is_weekend"] = df["order_date"].dt.dayofweek >= 5

# Task 3: price_band with np.where
df["price_band"] = np.where(df["unit_price"] >= 28000, "Premium", "Value")

# Task 4: check the new columns
print(df["is_weekend"].value_counts())
print(df["price_band"].value_counts())
```
